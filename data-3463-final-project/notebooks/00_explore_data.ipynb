# Exploring Data from Adzuna API call

Checking out if the raw data file matches with the actual API

```python
import json

with open("../data/raw/adzuna/adzuna_ca_toronto_data_analyst.json") as f:
    saved = json.load(f)

print(list(saved.keys()))
print(len(saved["results"]), "jobs in the file")
print("Result: '",saved["results"][12]["title"],"'")
```

Question: How many jobs list a salary?

```python
jobs = saved["results"]
with_salary = [job for job in jobs if job.get("salary_min")]
print(len(with_salary), "of", len(jobs), "jobs have a salary")
```

Checking the keys on one job

```python
print(sorted(saved["results"][0].keys()))
```

checking how much each keys show up

```python
from collections import Counter

counts = Counter(key for job in saved["results"] for key in job)
print(counts)
```

```python
flags = Counter(str(job["salary_is_predicted"]) for job in saved["results"])
print(flags)
```

```python
import math
import os

extra_calls = 0

for name in sorted(os.listdir("../data/raw/adzuna")):
    if name.startswith("adzuna_ca_") and name.endswith(".json"):
        with open(f"../data/raw/adzuna/{name}") as f:
            data = json.load(f)
        total = data["count"]
        pages = math.ceil(total / 50)
        extra_calls += pages - 1
        print(name, "|", total, "jobs |", pages, "page(s)")

print("extra calls needed for everything:", extra_calls)
```
